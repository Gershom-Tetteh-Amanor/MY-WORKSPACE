```python
import numpy as np

def arclength(f,a,b,n):
    x = np.linspace(a,b,n+1)
    y = f(x)
    dx = x[1]-x[0]
    dy = y[1]-y[0]
    L = np.sqrt(dx**2 + dy**2)
    for i in range(1,n):
        dx = x[i+1]-x[i]
        dy = y[i+1]-y[i]
        L += np.sqrt(dx**2 + dy**2)
    return L
```

```python
assert isclose(arclength(lambda x: x**2,1,3,100001),8.268145901058078)
```

```python
import numpy as np

def erf(x,n):
    # YOUR CODE HERE
    deltax = x/n
    xi = np.linspace(0,x,n+1)
    fxi = np.exp(-xi**2)
    return 2/np.sqrt(np.pi)*np.sum(fxi[:-1]*deltax)
    raise NotImplementedError()
```

```python
assert isclose(erf(1,1001),0.8430570034585734)
```

```python
assert isclose(erf(-5,101),-1.0279301774013976)
```

```python
assert isclose(erf(100,501),1.1126126913268979)
```

```python
def isDecreasing(x):
    # YOUR CODE HERE
    for i in range(len(x) - 1):
        if x[i] <= x[i + 1]:
            return False
    return True
    raise NotImplementedError()
```

```python
assert isDecreasing([9,3,0,-2,-9,-15]) == True
```

```python
assert isDecreasing([9,3,0,-2,-2,-15]) == False
```

```python
def unity(n):
    from cmath import sin, cos,pi
    return [cos((2*k*pi)/n)+1j*sin((2*k*pi)/n) for k in range(n)]
```
