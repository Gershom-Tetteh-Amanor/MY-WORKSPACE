Consider the mathematical structure with the coordinates (1:0; 0:0); (3:0; 5:2);
(􀀀0:5; 0:87); (􀀀6:0; 0:0); (􀀀0:5;􀀀0:87); (3:0;􀀀5:2): Write python code to find the
circumference of the structure. How would you extend it if your structure has many
points.

```python
def circumference(x,y):
    from math import sqrt
    D = 0.00
    for i in range(1,len(x)):
        di = sqrt((x[i]-x[i-1])**2 + (y[i]-y[i-1])**2)
        D += di
    return(D)
```

```python
def cross_prod(x,y):
    from math import acos,sqrt
    x_dot_y = x[0]*y[0]+x[1]*y[1]
    mag_x = sqrt((x[0])**2 + (x[1])**2)
    mag_y = sqrt((y[0])**2 + (y[1])**2)
    mag_prod = mag_x*mag_y
    res = acos(x_dot_y*mag_prod)
    print(res)
cross_prod([4,3],[5,12])
```

```python
def circumf(a,b,c,d,e,f):
    from math import sqrt
    AB = sqrt((b[0]-a[0])**2 + (b[1]-a[1])**2)
    BC = sqrt((c[0]-b[0])**2 + (c[1]-b[1])**2)
    CD= sqrt((d[0]-c[0])**2 + (d[1]-c[1])**2)
    DE= sqrt((e[0]-d[0])**2 + (e[1]-d[1])**2)
    EF = sqrt((f[0]-e[0])**2 + (f[1]-e[1])**2)
    FA= sqrt((a[0]-f[0])**2 + (a[1]-f[1])**2)
    circum = AB+BC+CD+DE+EF+FA
    return(circum)
```

A palindrome is a word, phrase, or sequence that reads the same backwards as
forwards, e.g. madam or nurses run. Write python code to test if a string is a
palindrome

```python
def pale(txt):
    mystr =txt.replace(' ','')
    mystr =mystr.lower()
    result = mystr[::-1] == mystr
    print('is '+str(txt)+' a palindrome? '+str(result))
    

```

```python
pale('madam')
pale('Nurses RuN')
pale('coconut')
pale('coccaine')
```
