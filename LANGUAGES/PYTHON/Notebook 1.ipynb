```python
x2 = np.linspace(-2,2,201)
s1 = np.sin(2*np.pi*x2)
s2 = np.sin(np.pi*x2)
s3 = np.sin(4*np.pi*x2)
plt.plot(x2,s1,label='sin(pi*x)')
plt.plot(x2,s2,label='sin(2*pi*x)')
plt.plot(x2,s3,label='sin(4*pi*x)')
plt.legend()
plt.savefig('plots.png')
```
