# Decision making

```python
a = [1,2,3,4,5,6,7,8,9]
i = 0
a_pow = []
while i < len(a):
    val = 10**-a[i]
    if val < 0.001:
        break
    a_pow.append(val)
    i += 1
    
a_pow    
```

```python
b = [1,2,3,4,5,6,7,8,9]
i = 0
b_pow = []
while i < len(b):
    val = 10**-b[i]
    if val < 0.001:
        continue 
    b_pow.append(val)
    i += 1
    
b_pow    
```

# Break and continue
## They both work with loops

```python
a = [2,3,8,9,0,2,4,1,10]
total = 0
for i in a:
   
    if i == 0:
        print('zero value was encountered, skipping......')
        #continue #skips and continue
        #break # terminates the loop
    print((i,(1/i)))   
    total += 1/i
```
