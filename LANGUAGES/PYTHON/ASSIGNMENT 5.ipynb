```python
from math import sqrt
```

```python
# Question 1.a
def pi_1(n):
    p = 0
    for k in range(0,n+1):
        p += (-1)**k/((3**k)*(2*k+1))
    return (6/sqrt(3))*p
```

```python
# Question 1.b
def pi_2(n):
    p = 0
    for k in range(0,n+1):
        p += (-3)**(-k)/((2*k+1))
    return (sqrt(12))*p
```

```python
# Question 1.c
def pi_3(n):
    p = 0
    for k in range(0,n+1):
        p += (-1)**k/((2*k+1))
    return 4*p
```
