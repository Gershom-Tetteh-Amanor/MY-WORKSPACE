A Pythagorean triple is the set of positive integers that satisfy the equation $$a^2+b^2=c^2.$$ Given two numbers $a$ and $b$ with $a<b$ we can say these numbers are components of a Pythagorean triplet if $\sqrt{a^2+b^2}$ or $\sqrt{b^2-a^2}$ is an integer. <br/><br/>
The two problems you will be solving are the same problem given different ways.

Write a function LastTrip(a,b) that returns the last component of the Pythagorean triplet as integer if $a,b$ are components of a Pythagorean triplet else returns nothing.

```python
def LastTrip(a,b):
    if a<b:
        from math import sqrt
        c = sqrt(a**2+b**2)
        if c-int(c) ==0:
            return(int(c))
        elif a>b:
            a,b =b,a
            c = sqrt(b**2-a**2)
            if c - int(c) ==0:
                return(int(c))
            else:
                return(None)
        
```
