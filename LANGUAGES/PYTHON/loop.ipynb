```python
# Consider the fxn g(x) = x^2 and f(x) = 2x-1 on [-2,5]
# write a code to approximate the max distance between the two functions
a = []

def g(x):
    return(x**2)
        
def f(x):
    return(2*x-1)
    
for x in range(-2,6):
   
    a.append(abs(g(x)-f(x)))
    #print(a)
max(a)
```

```python
#Consider a sequence defined by an=(1+1/n)^n.
# show that the sequence is convergent as n approaches infinity
# and find an approximation of the value it converges to

seq = []
def an(n):
    a = ((1 + 1/n)**n)
    return(a)
#seq.append(an(2))
#seq

for x in range(1,100000000000):
    y=abs(an(x))
    seq.append(y)
seq
 
max(seq)   

# the limit is approximated at 2.718268237028235
```
