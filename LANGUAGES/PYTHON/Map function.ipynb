# MAP

### Applies a function to each element in input sequence to give an output sequenc
### Each element in the output sequence can be traced back to the input with invers map


```python
list(map(lambda x: (x-3)/2, [5,7,9,11,13]))
```

```python
# we could create the sequence outside
a = [2,4,6,8,10]
f = (lambda x: (x-3)/2)
m = map(f,a)
list(m)
```

```python
list(map(lambda x: 2*x + 3, [1,2,3,4,5]))
```

```python
def add(a):
    return a+a
nums=(4,6,8,9)
out = map(add,nums)
result = list(out)
print(result)
```

```python
out = map(lambda a: a*a ,nums)
list(out)
```

```python
map_obj=map(str.upper, ['pen','book','phone'])
list(map_obj)
```

```python
mo=map(lambda x:len(list(x)),['pen','books','iphone','laptops'])
list(mo)
```

# Filter

### Apply function to all element and return sequences which are true with the condition

```python
def f(x):
    return x%2==1
f(2)
```

```python
(filter(f,[1,2,7,3,9,6,8]))
```

```python
list(filter(f,[1,2,7,3,9,6,8]))
```

```python
def f(a):
    if a>5:
        return a
###########################    
out = filter(f,(1,4,7,3,9))
list(out)
```

```python
out = filter(lambda a: (a>5),(1,4,7,3,9))
list(out)
```

```python
# Example with string operation
# Always the function must return simple True or false so thatthe filter return values that returns True
def vegi_with_e(x):
    return 'e' in x
out = filter(vegi_with_e, ['pears','garlic','ginger','beatroot','carrot'])
list(out)
```

```python
out=filter(lambda x: ('r' in x), ['pears','garlic','ginger','beatroot','carrot'])
list(out)
```

# Reduce Function

### It is an agregating function
### import it from the functools
### Returns a single value

```python
from functools import reduce
reduce(lambda x,y: max(abs(x),abs(y)),[1,-2,-7,-3,-9,6,8])
# in a pairwise manner, it takes first 2 elements and compare it and take the true and compare with the subsequent....
```

```python
reduce(lambda x,y : x*y, [4,9,2,3,8,1,5])
```
