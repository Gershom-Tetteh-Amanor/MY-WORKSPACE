As learnt in level 100, the nth root of unity is defined by the expression 
$$z_i = \cos(\frac{2k\pi}{n})+i\sin(\frac{2k\pi}{n}), \ \mathsf{with}\ k = 0,1,2,\ldots,n-1$$
Write a function runity(n) that returns the nth root of unity. Your function shoult return a list of the roots of unity.

```python
def runity(n):
    # YOUR CODE HERE
    from math import cos,sin,pi
    res = []
    k = range(0,n)
    for i in range(1,n+1):
        z = cos((2*k[i-1]*pi)/n) + 1j*sin((2*k[i-1]*pi)/n)
        res.append(z)
        i += 1
    return(res)    
    raise NotImplementedError()
```
