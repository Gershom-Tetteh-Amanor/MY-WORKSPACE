### Consider the problem of approximating the function $f (x) = 3x^2 - 5x + 6 $ with x in $[-1: 3]$: To approximate this we can create the discrete interval
$-1;-0:75;-0:5; ... ; 3$: Then for each element in x we apply the function. How can we generate such functions in python? How can we extend this to account for the piecewise function
f (x) = :((sin x)/x \for x!=0\else$
      

```python
# solution to the first part
def f(x):
    f = 3*x**2 - 5*x + 6
    return(f)

interval = list(range(-100,325,25))
[f(x/100) for x in interval]
```

```python
# solution to the second part
def sine(x):
    from math import sin
    f = (sin(x))/x
    return(f)
interval = list(range(-100,325,25))
[sine(x) if x!=0 else 1 for x in interval]
```
