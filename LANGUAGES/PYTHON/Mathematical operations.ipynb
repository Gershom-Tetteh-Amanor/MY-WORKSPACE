$\bullet$ The central difficulty in issuing $\underline{\textbf{\textit{Life insurance}}}$ is that of determining $\underline{\textbf{\textit{the time of death of the insured}}} \newline$
$\bullet$ Let time O denote the time of birth of an individual. We will assume everyone is alive at birth. $\newline$

$\bullet$ Let x be a continues and non negative random variable, Then x is an approximate random variable to model the age at which an individual dies and is therefore called $\underline{ \textbf{\textit{Age at Death}}} \newline$


The cumulative distribution of x is the probability that a new born dies before age x and is given by: 
$$F_o(x) = p(X \le x) = p(X<x)$$  $\newline$
$F_o(x)$ $\equiv$ The probability that death will occur prior to or at age x $\newline$
$$\bullet\ \  F_o(0) = 0 \ \ \ \ \ \ \ \ \ \ $$  
$$\bullet\ \lim_{x \rightarrow \infty} F_o(x) = 1$$
