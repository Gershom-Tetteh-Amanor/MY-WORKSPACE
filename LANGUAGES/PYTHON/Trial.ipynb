$Conside\r the\ function\ g(x)\ =\ x^2\ and\ f(x)\ =\ 2x-1\ on\ \left[-2,5\right]\ \ \ $
$write\ a\ code\ to\ approximate\ the\ max\ distance\ between\ the\ two\ functions$

```python
a = []
def g(x):
    return(x**2)       
def f(x):
    return(2*x-1)
for x in range(-2,6):
    a.append(abs(g(x)-f(x)))
print(a)
max(a)
```

```python
# A single function to find the maximum
def maximum(x):
    d =[]
    for i in range(len(x)):
        d.append(abs((x[i]**2)-(x[i]*2-1)))
    return max(d)
```

```python
assert maximum((1,2,3))==4
```

```python
assert maximum(range(-2,6))==16
```

```python
from math import log
def nth_position(an,a,r):
    n = ((log(an) - log(a))/(log(r))) + 1
    return (n)
```
