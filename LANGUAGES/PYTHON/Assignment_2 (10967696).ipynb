One of the integrals that is quite complicated is the error function $$\frac{2}{\sqrt{\pi}}\int_0^x e^{-t^2}dt=\mathrm{erf}(x).$$ It is however commonly used in probability, statistics and the theory of partial differential equations. We are going to approximate the error function using the Riemann sum. Using Riemann sums, we can approximate the definite integral $$\int_a^bf(x)\approx \sum_{i=1}^n f(x_i^*)\Delta x_i$$ where the interval $\left[ a,b \right]$ is partitioned into the points $a=x_0<x_1<x_2\ldots<x_n=b$ and $\Delta x_i=x_i-x_{i-1}$ and $x_i^*\in\left[ x_{i-1},x_i \right].$ This essentially approximates the integral using the sum of rectangles $$\sum R_i$$ where $R_i$ has height $f(x_i^*)$ and width $\Delta x_i.$ 

Write a function erf(x,n) that approximates the error function at the point $x$ using $n$ rectangles ($n+1$ points). We will choose $x_i^*$ to be the left endpoint $x_{i-1}.$ Remember to scale your result by $\frac{2}{\sqrt{\pi}}.$ Note: you can use linspace from numpy to perform the partitioning of the interval.

```python
import numpy as np

def erf(x,n):
    # YOUR CODE HERE
    deltax = x/n
    xi = np.linspace(0,x,n+1)
    fxi = np.exp(-xi**2)
    return 2/np.sqrt(np.pi)*np.sum(fxi[:-1]*deltax)
    raise NotImplementedError()
```

```python
def erf(x,n):
    from numpy import linspace
    from math import exp, pi,sqrt
    x = linspace(0,x,n+1)
    res = 0
    for i in range(x.size-1):
        res += exp(-x[i]**2)*(x[i+1]-x[i])
    return res*2/sqrt(pi)
```

Given the function $f(x),$ we can approximate the arc length from $(a,f(a))$ to $(b,f(b))$ as the sum $$\sum_{i=1}^n\sqrt{{\Delta x_i}^2+{\Delta y_i}^2}$$ where the interval $\left[ a,b \right]$ is partitioned into the points $a=x_0<x_1<x_2\ldots<x_n=b$ and $\Delta x_i=x_i-x_{i-1}$ and $\Delta y_i=f(x_i)-f(x_{i-1}).$ Write a function arclength(f,a,b,n) that takes f a function, the interval $[a,b]$, the number of subintervals n (or n+1 points) and returns an approximation of the arc length. You can use linspace from numpy to create your partition.
