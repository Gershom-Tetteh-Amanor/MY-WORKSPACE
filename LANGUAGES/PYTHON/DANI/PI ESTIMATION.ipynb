```python
def estimate_pi(r,n):
    import numpy as np

    theta = np.linspace(0,90,n+1)
    theta = np.radians(theta)
    x = r * np.cos(theta)
    y = r * np.sin(theta)

    delta_x = x[1:n+1]- x[0:n]
    delta_y = y[1:n+1]- y[0:n]

    li = np.sqrt(delta_x**2 + delta_y**2)
    
    C = np.sum(li)
    my_pi = 2*C/r
    return float(my_pi)
```

```python
r = 1
estimate_pi(r, n = 10)
```

```python
### Varying N
for i in range(1, 7):
    r = 1
    N = 10**i
    print("when n = " + str(N) + ": pi = " + str(estimate_pi(r, n=N)))
    #print(f"n = {N}, pi = {estimate_pi(r, n=N)}")
```

```python
### Varying r

r = list(range(1, 21))
for i in r:
    N = 1000
    print("when r = " + str(i) + ": pi = " + str(estimate_pi(r = i, n=N)))
    #print(f"n = {N}, pi = {estimate_pi(r, n=N)}")
```
