Write python function that finds the roots of a quadratic equation $ax^2+bx+c$ 

```python
def root_quad(a,b,c):
    D = b**2 -4*a*c
    x1 = (-b+ D**0.5)/2*a
    x2 = (-b- D**0.5)/2*a
    myroot = {"x1":x1,"x2":x2}
    return myroot
```

In algebra you learnt about the nth roots of unity, defined as
$$zi = \cos(\frac{2\pi k}{n}) + i \sin(\frac{2\pi k}{n})$$ , k = 0, 1, . . . , n − 1
Write code to find the values of the 5th roots of unity.

```python
def nth_unity(n):
    from math import cos,sin,pi
    z0 = complex(cos(2*pi*0/n) , sin(2*pi*0/n))
    z1 = complex(cos(2*pi*1/n) , sin(2*pi*1/n))
    z2 = complex(cos(2*pi*2/n) , sin(2*pi*2/n))
    z3 = complex(cos(2*pi*3/n) , sin(2*pi*3/n))
    z4 = complex(cos(2*pi*4/n) , sin(2*pi*4/n))
    return(z0,z1,z2,z3,z4) 
```

```python
nth_unity(5)
```

```python
nth_unity(5)
```

```python
def nth_root_of_unity(n):
    from math import sin, cos, pi
    for k in range(n):
        zi = complex(cos(2*pi*k/n),sin(2*pi*k/n))
        print(zi)
```

```python
nth_root_of_unity(5)
```
