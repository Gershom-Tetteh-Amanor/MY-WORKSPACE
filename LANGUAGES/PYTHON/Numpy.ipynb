# Functions of creating an array with numpy

### Evenly space

```python
a = np.arange(10) # 0 ... n-1
b  = np.arange(1,9,2) # 1...n-1 with steps of 2

a,b
```

### Number of points

```python
c = np.linspace(0,1,6) # start, end , num-points
d = np.linspace(0,1,5,endpoint=False)
c,d
```

### Common arrays

```python
a = np.ones((3,3))
b = np.zeros((2,3))
c = np.eye(3)
d = np.diag(np.array([1,2,3,4]))
a,b,c,d
```

### np.random: random numbers (Mersenne Twister PRNG):

```python
a = np.random.rand(4) # Uniform in [ 0,1]
b = np.random.randn(4) #Gaussian
c = np.random.seed(1234) # Setting the seed
a,b,c
```
