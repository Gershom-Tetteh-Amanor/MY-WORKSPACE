```python
def anagornal(u,v):
    from math import sqrt
    p=[]
    if len(u)==len(v):
        for i in range(len(u)):
            mag_uv= sqrt((u[i]-v[i])**2)
            prod = u[i]*v[i]
            if mag_uv != 0:
                an =(prod/mag_uv**2)*u[i]
                p.append(round(an,3))
    return (tuple(p))             
```

```python
anagornal([1,-2,3,4,8,100],[5,6,7,8,2,-3])
```

```python
anagornal((1,3,5,7),[2,4,6,8])
```
