```python
# A function to check if a number is a prime number
def isp(x):
    import numpy as np
    if (x>=0) and (x<=1):
        print(x,' is not prime')
    elif x == 2:
        print(x, ' is prime')
    elif x>2:
        pc = []
        i = list(range(2,x))
        p = list(np.ones(x-2, dtype=int)*x)
        for j in range(x-2):
            pc.append(p[j]%i[j])
        if ((0 in pc)==True):
            print(x,' is not prime')
        else:
            print(x,' is prime')
    else:
        print('consider a non negative argument')
```

```python
isp(571)
```

## ODD NUMBERS

```python
# Function to check if a number is odd
def isod(x):
    if x%2 !=0:
        print(x,' is odd')
    else:
        print("😎")    
```

```python
isod(23)
```

```python
# Function to return logical values; true if it is odd
def isodb(x):
    if x%2 !=0:
        return True
    else:
        return False
```

```python
# Function to print the odd numbers within a specified range
def nodd(a,b):
    x = range(a,b+1)
    od = []
    for i in range(len(x)):
        if isodb(x[i])==True:
            od.append(x[i])
    print('The odd numbers from',a,' and ',b,' are :',od)
    print("There are ", len(od),' odd numbers from ',a,'to',b)        
```

```python
nodd(-9,9)
```

## EVEN NUMBERS

```python
# Function to check if a number is even
def isev(x):
    if x%2 ==0:
        print(x,' is even')
    else:
        print("😎")    
```

```python
isev(12)
```

```python
# Function to return logical values; true if it is even
def isev(x):
    if x%2 ==0:
        return True
    else:
        return False
```

```python
# Function to print the even numbers within a specified range
def nev(a,b):
    x = range(a,b+1)
    od = []
    for i in range(len(x)):
        if isev(x[i])==True:
            od.append(x[i])
    print('The even numbers from',a,' and ',b,' are :',od)
    print("There are ", len(od),' even numbers from ',a,'to',b)        
```

```python
nev(-100,100)
```
