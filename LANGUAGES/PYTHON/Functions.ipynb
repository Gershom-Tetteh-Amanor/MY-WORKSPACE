```python
# A function to calculate the roots of a quadreatic function
def quadratic(a,b,c):
    D=b**2-4*a*c #The discriminant
    x1=(-b+D**0.5)/(2*a)
    x2=(-b-D**0.5)/(2*a)
    roots={"X1":x1,"X2":x2}
    return(roots)
```

If the nth root of unity is defined as :
$\ \ Z_i\ = \cos{\frac{2k\pi}{n}} + i\sin{\frac{2k\pi}{n}}\ \  \ ;\ k=0,1,2, \dots,n-1\ \ \ \ $
write a function to find the 5th root of unity

```python
from math import cos, sin, pi
def fithroot(n):
    if n == 5:
        k=n-1
        i=n
        zi = cos((2*k*pi)/n) + i*sin((2*k*pi)/n)
        return zi

fithroot(5)
```
