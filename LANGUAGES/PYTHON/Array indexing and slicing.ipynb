# Mask array

```python
x=np.linspace(-2,5,101)
y = np.zeros_like(x)
```

```python
y[x<=0] = np.sin(x[x<=0])
y[x>=0] = np.sin(x[x>0]*3)
```

```python
print(y,x)
```

```python
# Importing the matplob 
import matplotlib.pyplot as plt
```

```python
plt.plot(x,y)
```
