```python
from math import (pi,sqrt)
def radius(A):
    r = sqrt(A/pi)
    return(r)

radius(1000)
```

# QUESTION 2
## A palindrome is a word, phrase, or sequence that reads the same backwards as
## forwards, e.g. madam or nurses run. Write python code to test if a string is a
## palindrome

```python
def ispalindrome(Text):
    ans = Text[::-1]
    if ans == Text:
        print("Yes ", Text, "is a Palindrome")
    else:
        print("Sorry, ", Text, " is not a palindrome")
    return(Text,ans)
    
```

```python
ispalindrome('nurses run')
```
