```python
#Consider the mathematical structure with the coordinates (1.0, 0.0); (3.0, 5.2);
#(-0.5, 0.87); (-6.0, 0.0); (-0.5,-0.87); (3.0,-5.2): Write python code to find the
#circumference of the structure. How would you extend it if your structure has many points.
```

```python
# Assuming the above cordinates are the cordinates of circldes centered at the origin
# then we can write this function to calculate the circumference
from math import (sqrt,pi)
def circum(x,y):
    rsq=x**2 + y**2
    r = sqrt(rsq)
    cf = 2 * pi * r
    return(cf)
```
