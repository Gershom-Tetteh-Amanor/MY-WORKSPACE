Python supports the concept of nested lists. One potential use of this concept is as matrices. For instance the following code
```
A = [[a,b],[c,d]]
```
can represent the matrix $$A=\begin{bmatrix} a & b \\ c & d \end{bmatrix}.$$ It is therefore easy to see that 
```
A[0][0]=a,
A[0][1]=b,
A[1][0]=c,
A[1][1]=d
```

Given the matrix $$A=\begin{bmatrix} a & b \\ c & d \end{bmatrix}$$ we know that the determinant of $A$ is given as $$\det(A)=ad-bc.$$ Write a function det(A) that takes as input A a 2x2 matrix written as a list (nested) and returns the determinant of the matrix represented by the list A. Your code should ensure that your matrix is 2x2 and return none if that condition is violated

````python
def det(A):
    # YOUR CODE HERE
    # we check if each of the nested lists and the external list are of length 2
    if len(A)==2:
        D =(A[0][0]*A[1][1])-(A[0][1]*A[1][0])
    else:
        D = None
    return(D)
    raise NotImplementedError()
````

````python
assert isinstance(det([[1,0],[0,1]]),int)
````

````python
assert det([[0,1],[1,0]]) == -1
````

````python
assert det([[5,0],[2,4]]) == 20
````

````python
A = [[4,5],[-8,5]]
assert det(A) == A[0][0]*A[-1][-1] - A[0][-1]*A[-1][0]
````

The transpose of the matrix $$A=\begin{bmatrix} a & b \\ c & d \end{bmatrix}$$ is the matrix $$A=\begin{bmatrix} a & c \\ b & d \end{bmatrix}.$$ Write a function T(A) that takes as input a matrix A written as a nested list and returns the transpose as a nested list. 

````python
def T(A):
    # YOUR CODE HERE
    if len(A)==2:
        T =[[A[0][0],A[1][0]],[A[0][1],A[1][1]]]
    else:
        T = None
    return(T)
    raise NotImplementedError()
````

````python
assert T([[1,5],[2,3]]) == [[1, 2], [5, 3]]
````

````python
assert T([[7,-1],[5,4]]) == [[7, 5], [-1, 4]]
````

We can find the inverse of a 2x2 matrix $$A=\begin{bmatrix} a & b \\ c & d \end{bmatrix}$$ as $$A^{-1}=\begin{bmatrix} \frac{d}{det(A)} & \frac{-b}{det(A)} \\ \frac{-c}{det(A)} & \frac{a}{det(A)} \end{bmatrix}.$$ Write a function inv(A) that takes as input a 2x2 matrix A written as a list (nested) and returns its inverse as a nested list.

````python
def inv(A):
    # YOUR CODE HERE
    dt = det(A)
    #if len(A)==2:
    I=[[A[1][1]/dt,-1*A[0][1]/dt],[-1*A[1][0]/dt,A[0][0]/dt]]
    #else:
    #   I = None
    return(I)    
    raise NotImplementedError()
````

````python
assert abs(det(inv([[4,5],[-8,5]]))*det([[4,5],[-8,5]])-1)<1e9
````

````python
assert inv([[1,0],[0,1]]) == [[1,0],[0,1]]
````

````python
assert inv([[0,1],[1,0]]) == [[0.0, 1.0], [1.0, 0.0]]
````

Given the 2x2 matrix $$A=\begin{bmatrix} a_{11} & a_{12} \\ a_{21} & a_{22} \end{bmatrix}$$ and the vector $\boldsymbol{b}$, $$B=\begin{bmatrix} b_{1}\\ b_{2} \end{bmatrix},$$ the product $A\boldsymbol{b}$ is given as
$$A\boldsymbol{b}=\begin{bmatrix} a_{11}b_{1}+a_{12}b_{2} \\ a_{21}b_{1}+a_{22}b_{2} \end{bmatrix}.$$ 
Write a function mvprod(A,b) that takes as input a 2x2 matrices A written as nested list, and a two element vector b written as a simple list and returns their product as a simple list.

````python
def mvprod(A,b):
    # YOUR CODE HERE
    prod = [A[0][0]*b[0] + A[0][1]*b[1], A[1][0]*b[0] + A[1][1]*b[1]]
    return(prod)
    raise NotImplementedError()
````

````python
assert mvprod([[1,2],[1,4]],[2,3]) == [8, 14]
````

````python
assert mvprod([[2,1],[3,0]],[4,1]) == [9, 12]
````

Given the 2x2 matrices $$A=\begin{bmatrix} a_{11} & a_{12} \\ a_{21} & a_{22} \end{bmatrix}$$ and $$B=\begin{bmatrix} b_{11} & b_{12} \\ b_{21} & b_{22} \end{bmatrix},$$ the product $AB$ is given as
$$AB=\begin{bmatrix} a_{11}b_{11}+a_{12}b_{21} & a_{11}b_{12}+a_{12}b_{22} \\ a_{21}b_{11}+a_{22}b_{21} & a_{21}b_{12}+a_{22}b_{22} \end{bmatrix}.$$ 
Note that $AB$ can also written as $$AB = \begin{bmatrix} A\boldsymbol{b}_1 & A\boldsymbol{b}_2 \end{bmatrix}$$ where $$\boldsymbol{b}_k = \begin{bmatrix} b_{1k} & \ldots & b_{nk} \end{bmatrix}^T$$
Write a function mprod(A,B) that takes as input two 2x2 matrices A, B written as nested lists and returns their product as a nested list.

````python
def mprod(A,B):
    # YOUR CODE HERE
    prod = [[A[0][0]*B[0][0] + A[0][1]*B[1][0], A[0][0]*B[0][1] + A[0][1]*B[1][1]], [A[1][0]*B[0][0] + A[1][1]*B[1][0], A[1][0]*B[0][1] + A[1][1]*B[1][1]]]
    return(prod)
    raise NotImplementedError()
````
