Consider the mathematical structure with the coordinates (1.0, 0.0); (3.0, 5.2);
(-0.5, 0.87); (-6.0, 0.0); (-0.5,-0.87); (3.0,-5.2): Write python code to find the
circumference of the structure. How would you extend it if your structure has many
points.

```python
from math import sqrt
import numpy as np
def circ(C):
    cr=0
    for i in range(len(C)-1):
        for j in range(1):
            cr+=(C[i][j]-C[i][j+1])**2
    return print('The circumference of the structure is ', sqrt(cr), ' units')        
```

```python
circ([(1,0),(3,5.2),(-0.5,0.87),(-6.0,0.87),(-0.5,-0.87),(3.0,-5.2)])
```

A palindrome is a word, phrase, or sequence that reads the same backwards as
forwards, e.g. madam or nurses run. Write python code to test if a string is a
palindrome

```python
def palindrome(Text):
    t=Text.lower() # convert all to same case
    tl=t.split() # makes it as a list at the spaces
    tt="".join(tl) # puting the listed elements together
    for i in range(len(tt)):
        if tt[i]!=tt[-(i+1)]:
            return False
    return True    
```

```python
assert palindrome('step on no pets')==True
```

```python
assert palindrome('Nurses run')==True
```

```python
assert palindrome('madam')==True
```

```python
assert palindrome('nurse run')==False
```

To convert a decimal integer to a binary string, we perform successive divisions
until the quotient is equal to 0 and read the remainders backwards to get the
binary string. Example, to convert the number 25 to binary we have 12r 1, 6r 0, 3r 0, 1r 1,0r 1

```python
def binary(n):
    whole=[]
    remainder=[]
    bina=[]
    if n//2!=0:
        whole.append(n//2)
        remainder.append(n%2)
        for i in range(10000):
            if whole[i]==0:
                break
            whole.append(whole[i]//2)
            remainder.append(whole[i]%2)
        for j in range (1,len(remainder)+1):
            bina.append(str(remainder[-j]))
            f=''.join(bina)
    return int(f)
```

```python
assert binary(768)==1100000000
```

```python
assert binary(265)==100001001
```

Related to the previous example is the problem of converting a binary number to a
decimal one. In this case we find the sum of the successive multiplication of the
elements of th binary string by increasing powers of 2. Thus
$11001_2 = 1\times{2^4} + 1\times{2^3} +0\times{2^2} + 0\times{2^1} +1\times{2^0} = 16+8+0+0+1=25 $: Write
a python function that converts a binary string to a decimal. Test your function on
the numbers 100001001; 01010101010101 and 101100101:

```python
def decimal(n):
    num = str(n)
    x=0
    for i in range(len(num)):
        y = range(len(num)-1,-1,-1)
        x += int(num[i])*2**(y[i])
        i += 1    
    return x
```

```python
assert decimal(100001001)==265
```

```python
assert decimal(1010101010101)==5461
```

```python
assert decimal(101100101)==357
```

Given a sequence $x_1 , x_2 ,x_3 , \dots , x_{n-1},x_n$ the sequence is a geometric sequence if
$\frac{x_i}{x_{i-1}} = k\ ;\ \ i = 2,3,4\dots , n:$ Write a function that takes a list of numbers and
determines if the list represents a geometric sequence. Test your function with the
sequence (4, 15, 26, 37, 48, 59, 70) and (1,-3,9,-27,81,-243):

```python
def geometric(x):
    k = x[1]/x[0]
    n = len(x)
    for i in range(2,n):
        if x[i]/x[i-1] !=k:
            return False
    return True        
```

```python
geometric((4, 15, 26, 37, 48, 59, 70))
```

```python
geometric((1,-3,9,-27,81,-243))
```

The vector equation of a line is given by $r = a + td$ : This can be expanded as the
equation $(x; y) = (a1; a2) + t(d1; d2)$ where $a$ is a point on the line and $d$ is the
direction vector. Write a function that takes $a\ and\ d$ as tuples and $t$ as a float and
determines the position vector of the point

```python
def divector(a,d,t):
    r=[]
    if (type(a)==tuple) and (type(d)==tuple):
        for i in range(len(a)):
            r.append(a[i]+t*d[i])
        return tuple(r)
    else:
        return print('InputError: divector takes two tuples and a float as arguments')
```

```python
divector((4,3,1),(5,12,1),2.0)
```

Let $a = (a1; a2)\ and\ b = (b1; b2)$ be vectors, we know that $a.b = |a|.|b| \cos \theta$:
Find the angle between the two vectors defined by the tuple (4,3) and (5,12).

```python
def angle(a,b):
    from math import acos, sqrt
    if (type(a) and type(b))==tuple:
        ab=0
        mag_a=0
        mag_b=0
        for i in range(len(a)):
            ab += (a[i]*b[i])
            mag_a += (a[i]**2)
            mag_b += (b[i]**2)
        mag_prod = (sqrt(mag_a)*sqrt(mag_b))    
        return acos(ab/mag_prod)
    else:
        return print('InputError: angle elements must be two tuples')    
```

```python
# Alternative code without a loop
def ang(a,b):
    if (type(a)==tuple) and (type(b)==tuple):
        import numpy as np
        from math import sqrt, acos
        a = np.array(a)
        b = np.array(b)
        ab=sum(a*b)
        mag_a = sqrt(sum((a)**2))
        mag_b = sqrt(sum((b)**2))
        return acos(ab/(mag_a*mag_b))
    else:
        return print('InputError: angle elements must be two tuples')
```

Write a function for finding the determinant of a 2 $\times$ 2 matrix $A\ =\ \begin{bmatrix} a&b \\ c&d \end{bmatrix}$
given as a nested list [[a,b],[c,d]]

```python
# Assuming the matrx is given as a nested list
def det(A):
    ad=A[0][0]*A[1][1]
    bc=A[0][1]*A[1][0]
    return ad-bc
```

```python
assert det([[2,1],[1,0]])==-1
```

```python
assert det([[1,0],[0,1]])==1
```

In algebra you learnt about the nth roots of unity, defined as
$z_i = \cos\frac{2k\pi}{n}+i\sin\frac{2k\pi}{n}\ ;\ k=0,1,2,\dots,\ n-1$
Write code to find the values of the 5th roots of unity.

```python
def unity(n):
    from cmath import sin, cos,pi
    return [cos((2*k*pi)/n)+1j*sin((2*k*pi)/n) for k in range(n)]
```
