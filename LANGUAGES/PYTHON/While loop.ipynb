# In how many years an investment of GHS 20000 will double if interest is 
# calculated under compound interest at a rate of 5%


```python
# In how many years an investment of GHS 20000 will double if interest is 
# calculated under compound interest at a rate of 5%

principal = 20000
amount = 20000
years = 1
rate = 0.05
while (amount <= 2*principal):
    amount = principal*(1+rate)**years
    print('Afer ' ,years,'yr, The Amount is GHS',round(amount,2))
    years +=1
```

```python
# How many tosses are required to get series of sixes on both the faces 
#if two isx faced dice🎲 aer tossed together?

from random import sample
# sample(range(1,7),1)[0] a code that helps to sample and print a number not list

s=2 # s is 2 because that is the min value when you toss 2 dice together
tosses = 0
while(s != 12):
    n1,n2 = sample(range(1,7),1)[0],sample(range(1,7),1)[0]
    s = n1 + n2
    tosses += 1
print(tosses)    
```
